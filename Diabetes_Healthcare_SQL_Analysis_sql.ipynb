```python
-- KPI 1
-- Total Patients
create database diabetes_prediction;
use diabetes_prediction;
select * from diabetes_data
limit 10;
-- Total Patients
select count(round_age) as total_patients
from diabetes_data;
-- Total Diabetic Patients
select count(Diabetes_category) as total_diabetic_patients
from diabetes_data
where Diabetes_category = "Diabetic";
-- Diabetes Percentage
select (
(select count(Diabetes_category) as total_diabetic_patients
from diabetes_data
where Diabetes_category = "Diabetic")
* 100.0 /
(select count(round_age) as total_patients
from diabetes_data)
) as diabetic_percentage;
-- Average Age of Patients
select avg(round_age) as average_age
from diabetes_data;
-- Average BMI of Patients
select avg(bmi) as average_bmi
from diabetes_data;
-- Average Blood Glucose Level
select avg(blood_glucose_level) as average_blood_glucose_level
from diabetes_data;
-- Average HbA1c Level
select avg(HbA1c_level) as Average_HbA1c_Level
from diabetes_data;
-- Does Obesity Increase Diabetes Risk?
select BMI_CATEGORY,Diabetes_category,count(*) as patient_count
from diabetes_data
group by BMI_CATEGORY,Diabetes_category;
-- Does Smoking Increase Diabetes Prevalence?
select smoking_history,Diabetes_category,count(*) as smoking_diabetic_count
from diabetes_data
group by smoking_history,Diabetes_category;
-- Does Hypertension Contribute to Diabetes?
select hypertension_category,Diabetes_category,count(*) as Hypertension_diabetic_count
from diabetes_data
group by hypertension_category,Diabetes_category
having hypertension_category = "Hypertensive";
-- Does Heart Disease Correlate with Diabetes?
select Diabetes_category,count(*) as Heart_Disease_diabetic_count
from diabetes_data
group by heart_disease,Diabetes_category
having heart_disease = "1";
-- Which Age Group Has Highest Diabetes Prevalence?
select AGE_GROUP,Diabetes_category,count(*) as AGE_GROUP_diabetic_count
from diabetes_data
group by AGE_GROUP,Diabetes_category;
-- Are Males or Females More Diabetic?
select gender,Diabetes_category,count(*) as gender_diabetic_count
from diabetes_data
group by gender,Diabetes_category;
-- Which Age Group Has Highest Average Glucose Level?
select AGE_GROUP,
 round(avg(blood_glucose_level),2) as average_glucose_level
FROM diabetes_data
group by AGE_GROUP
order by average_glucose_level desc;
-- Which Age Group Has Highest Average HbA1c Level?
select AGE_GROUP,
 round(avg(HbA1c_level),1) as average_HbA1c_level
FROM diabetes_data
group by AGE_GROUP
order by average_HbA1c_level desc;
-- Which Smoking Category Has Highest Average Glucose Level?
select smoking_history,
  round(avg(blood_glucose_level),2) as average_glucose_level1
FROM diabetes_data
group by smoking_history
order by average_glucose_level1 desc;
-- Which BMI Category Has Highest Average HbA1c Level?
select BMI_CATEGORY,
round(avg(HbA1c_level),1) as average_HbA1c_level1
FROM diabetes_data
group by BMI_CATEGORY
order by average_HbA1c_level1 desc;
-- Combined Impact of Obesity and Hypertension on Diabetes
select BMI_CATEGORY,hypertension_category,Diabetes_category,count(*) as impact_obese_HTN_on_diabetes
from diabetes_data
group by BMI_CATEGORY,hypertension_category,Diabetes_category
having BMI_CATEGORY = "Obese" and
hypertension_category = "Hypertensive"
;
```
